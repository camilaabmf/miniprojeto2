## 2. Limpeza

Deixe cada tabela num estado em que dá para confiar.

- Datas e horas: viram `datetime`? Vale montar uma coluna única de data + hora?
- Decimais: o site usa vírgula. Isso já foi resolvido na Etapa 1 ou sobrou aqui?
- O que fazer com nulos: descartar a linha, preencher, ou manter e documentar?
- Duplicatas: existem? De onde vieram?

```python
def criar_data_hora(ano, dia_bruto, hora_bruto):

    dia = dia_bruto.str.split().str[0].astype(int)
    mes = dia_bruto.str.split().str[1].map(MESES)
    hora = limpa_hora(hora_bruto)

    data = pd.DataFrame({
    "year": ano,
    "month": mes,
    "day": dia,
    "hour": hora
    })

    return pd.to_datetime(data)

def limpa_hora(hora):

    if hora.str.contains(":").all():
        return hora.str.split(":").str[0].astype(int) 
    elif hora.str.contains("h").all():
        return hora.str.replace("h", "").str.strip().astype(int)
    else:
        raise ValueError("Formato desconhecido")

def limpar_coeficiente_mares(coluna_coef):
    coef_classific = coluna_coef.str.split(n=1)
    return coef_classific.str[0].astype(int)

def limpar_classificacao_mares(coluna_coef):
    coef_classific = coluna_coef.str.split(n=1)
    return coef_classific.str[1]

def limpar_altura_mares(coluna_mare):
    altura_bruta = coluna_mare  .str.split().str[1]
    return altura_bruta.str.replace(",", ".").astype(float)
    
def limpar_data_mares(data, hora_bruta):
    
    if data.str.contains("'").all():
        data_limpa = data.str.split("'").str[1]
    else:
        raise ValueError("Formato desconhecido")

    ano = data_limpa.str.split("-").str[0].astype(int)
    mes = data_limpa.str.split("-").str[1].astype(int)
    dia = data_limpa.str.split("-").str[2].astype(int)

    horas_minutos = hora_bruta.str.split(":")
    horas = horas_minutos.str[0].astype(int)
    minutos = horas_minutos.str[1].astype(int)
    
    data = pd.DataFrame({
    "year": ano,
    "month": mes,
    "day": dia,
    "hour": horas,
    "minute": minutos
    })

    return pd.to_datetime(data)
```

```python
#Limpeza de ondas.csv

ondas = ondas_bruto.copy()          

#limpeza de altura:
altura = ondas["altura_onda(m)"]
altura = altura.str.replace(",", ".")
altura = altura.str.replace("m", "")
ondas["altura_onda(m)"] = altura.astype(float)

ondas["data_hora"] = criar_data_hora(ondas["ano"], ondas["dia"], ondas["hora_dia"])

ondas = ondas[["data_hora", "direcao_onda", "altura_onda(m)"]]    
```

## 4. Juntar as tabelas

Esta é a parte difícil, e ela tem mais de uma resposta certa.

As granularidades não batem: onda e vento são leituras horárias; maré são ~4
eventos por dia (os instantes de máxima e mínima). Para ter uma linha por hora
com as três informações você precisa decidir:

- Como levar a maré para a grade horária? Interpolar entre eventos? Repetir o
  último valor? Guardar apenas se a maré está subindo ou descendo? O que cada
  escolha assume sobre a curva real da maré — e onde essa suposição erra?
- Qual o tipo de junção? A previsão de onda e a de vento cobrem exatamente as
  mesmas horas? Se não, você mantém as horas incompletas ou corta?
- Que variáveis derivadas de tempo ajudam a responder a pergunta do projeto
  (dia da semana, período do dia, ...)?

Justifique as escolhas em markdown. Elas são a base das duas próximas etapas.

```python
dataset = (
    ondas.merge(vento, on="data_hora", validate="one_to_one")
         .merge(chuva, on="data_hora", validate="one_to_one")
         .sort_values("data_hora")
         .reset_index(drop=True)
)

mares_ord = mares.sort_values("data_hora")
t0 = dataset["data_hora"].min()

t_eventos = (mares_ord["data_hora"] - t0).dt.total_seconds().to_numpy()
t_horas = (dataset["data_hora"] - t0).dt.total_seconds().to_numpy()

dataset["altura_mare_interp(m)"] = np.interp(
    t_horas, t_eventos, mares_ord["altura_mare(m)"].to_numpy()
)

variacao = np.gradient(dataset["altura_mare_interp(m)"].to_numpy())
dataset["mare_sentido"] = np.where(variacao > 0, "subindo", "descendo")

dataset["data"] = dataset["data_hora"].dt.normalize()
coef_dia = (
    mares.assign(data=mares["data_hora"].dt.normalize())
         [["data", "coeficiente", "classificacao_coef"]]
         .drop_duplicates("data")
)
dataset = dataset.merge(coef_dia, on="data", how="left", validate="many_to_one")

dataset["hora"] = dataset["data_hora"].dt.hour
dataset["periodo"] = pd.cut(
    dataset["hora"], [-1, 5, 11, 17, 23],
    labels=["madrugada", "manhã", "tarde", "noite"],
)                                    

display(dataset[dataset["data_hora"].dt.day == 9][
    ["data_hora", "altura_mare_interp(m)", "mare_sentido", "altura_onda(m)", "velocidade"]])

repetida = pd.merge_asof(dataset[["data_hora"]].sort_values("data_hora"),
                         mares.sort_values("data_hora")[["data_hora", "altura_mare(m)"]],
                         on="data_hora")["altura_mare(m)"].to_numpy()
erro = np.abs(repetida - dataset.sort_values("data_hora")["altura_mare_interp(m)"].to_numpy())
print("media de erro e erro máximo:", erro.mean(), erro.max())
```

## Decisões da junção

**Maré** - Eu decidi fazer uma interpolação linear no tempo entre eventos consecutivos, visto que a maré tem, na maioria das vezes, 4 eventos por dia e as outras tabelas têm uma leitura por hora, com a tábua do mês inteiro para que as horas das pontas tenham um evento antes e outro em seguida. Isso assume que a água sobe ou desce a velocidade constante entre as marés, de forma que a curva real é mais suave, o que faz com que o erro seja maior ao se aproximar da alta e da baixa. É importante salientar que o erro máximo estimado é de cerca de 0,2 m, o que é uma estimativa teórica, e não uma medição. O sentido da maré (`mare_sentido`) vem da variação da altura interpolada. Optei por não repetir o último valor, pois isso faria com que a altura ficasse parada até o próximo evento e erraria em até cerca de 2 m, e guardar só o sentido (subindo ou descendo) faria com que a magnitude fosse perdida.

**Coeficiente** - Juntei pela data porque na madrugada a última maré seria a do dia anterior.

**Junção** - Ondas, vento e chuva foram juntados por `data_hora` com `inner`. As três tabelas cobrem as mesmas 168 horas, então nenhuma hora foi cortada.

**Chuva** - A probabilidade de chuva ficou entre 4% e 43%, e a chuva prevista é fraca, com no máximo 0,4 mm por hora. Como esse elemento quase não diferencia as horas nessa semana, eu optei por não o usar na regra de classificação, mas mantive as colunas no dataset. 

**Variáveis derivadas** - Criei as variáveis `hora`, `periodo` (madrugada, manhã, tarde e noite) e `mare_sentido` para me auxiliar nas análises desses dados. Optei por não usar o dia da semana, porque cada dia já aparece uma vez na janela.
